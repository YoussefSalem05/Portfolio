```python
#Question 2: Write a Python program that takes two numbers as input and prints their sum
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))
sum = num1 + num2
print("The sum of", num1, "and", num2, "is:", sum)
```

```python
#Question 3: Write a Python program that asks the user for a number and checks if it is even or odd
number = int(input("Enter a number: "))
print("The number is even" if number % 2 == 0 else "The number is odd")
```

```python
#Question 4: Write a Python program that prints numbers from 1 to 10 using a for loop
for i in range(1, 11):
    print(i)
```

```python
#Question 5: Write a Python program that asks the user for a number and prints the numbers from 1 to N
N = int(input("Enter a number: "))
assert N > 0, "Please enter a positive integer"
for i in range(1, N + 1):
    print(i)
```

```python
#Question 6: Write a Python program that asks the user for a word and prints the number of characters in that word
word = input("Enter a word: ")
print(f"The number of characters in '{word}' is: {len(word)}")
```

```python
#Question 7: Write a Python program that creates a list of 5 numbers and prints the largest number in the list
import random as rnd
numbers = [rnd.randint(1, 100) for _ in range(5)]
print("The list of numbers is:", numbers)
print("The largest number in the list is:", max(numbers))
```
