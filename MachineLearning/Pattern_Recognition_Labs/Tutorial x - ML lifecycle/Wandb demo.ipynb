# Tracking a PyTorch Experiment with Weights & Biases

This demo logs loss and accuracy from a small PyTorch model. It uses the `WANDB_API_KEY` environment variable for authentication and sends run metrics to the configured Weights & Biases project.

```python
import torch
import torch.nn as nn
import torch.optim as optim
import wandb
import os
```

```python
wandb.login()
wandb.init(project="wandb_demo_project", name="dummy_training_run") # project name and run name
```
